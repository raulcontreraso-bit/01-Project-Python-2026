```python
#
# Pipeline: 1.- Conjunto de datos sin procesar(raw datasets) --> DataFrame inicial
#
# Paso 1— Importar todas las librerias necesarias
#
import pandas as pd  # tables (like Excel inside Python)
import numpy as np   # numbers and random generation
#-----------------------------------------------------------------------------------  

# Paso 2 — Reproducibility {reproducibilidad,ejecuta el mismo código bajo las mismas condiciones, y obtiene los mismos resultados. }
           #Reproducibilidad del experimento, no del resultado clínico.
np.random.seed(42) #Genera los mismos numero aleatorios.”

# Paso 3 — Cuantos Pacientes se generaran de forma sintetica
n_patients = 20    #  El alcance de la muestra

#Paso 4 — Diccionario de Datos (Data dictionary)
data = { 
     # Paso 4.1 — Genera numeros aleatorios de marcadores medicos 
     #uniform(a, b) → Genera numeros aleatorios dentro de un rango(flat distribution between a and b)
     
    #Generate n_patients
   "patient_id": np.arange(1, n_patients + 1), 
   "patient_name": [f"Patient_{i}" for i in range(1, n_patients + 1)], 
      
   "blood_sugar_mg_dl": np.random.uniform(70, 200, n_patients),  
   "max_heart_rate_bpm": np.random.uniform(90, 190, n_patients), 
     
   "total_cholesterol_mg_dl": np.random.uniform(120, 300, n_patients), 
   "ldl_cholesterol_mg_dl": np.random.uniform(60, 220, n_patients), 
   "hdl_cholesterol_mg_dl": np.random.uniform(30, 100, n_patients), 
   "triglycerides_mg_dl": np.random.uniform(50, 400, n_patients), 
   "lipoprotein_a_mg_dl": np.random.uniform(5, 150, n_patients), 

   #Paso 4.2 —  Columnas binarias, representan estados SI/NO (Binary columns 
   #choice) “For each patient, randomly pick either 'yes' or 'no'.”
   "diabetes": np.random.choice(["yes", "no"], n_patients), 
   "smoker": np.random.choice(["yes", "no"], n_patients), 
   } 

      
#Paso 5 — Crea la matriz tabular numerica
    #Building the DataFrame: transform  the dictionary of columns → turn it into a table.”  #rows = patients, #columns = features
    #El modelo ML requiere de una matriz numérica en forma tabular (filas × columnas). El DataFrame transforma el diccionario a una forma tabular. 

df = pd.DataFrame(data) 

#Paso 6 —   filtra el DataFrame y se queda solo con las columnas numéricas (int, float: edad, colesterol, presión, etc.).
            #.columns: extrae los nombres de esas columnas, no los valores. numeric_cols es una lista de nombres de columnas numéricas.
            #luego aplicar el StandardScaler solo a estas columnas.{el artículo llama: data preprocessing, normalization, feature preparation
numeric_cols = df.select_dtypes(include="number").columns  
 
  

#Paso 7 — Redondeo(Rounding)
         # Opcional: para mejor vizualizacion (readability), menos decimales, mas facil de leer
df[numeric_cols] = df[numeric_cols].round(2) 


#Paso 8 — Grabar el Archivo denerado de datos sinteticos
          # Save to Excel 
df.to_excel("synthetic_clinical_data.xlsx", index=False) 


#Paso 9  — Despliega la tabla
           #Muestra las 5 filas, la estructura, y los tipos de datos
df.head() 
```

```python
# 
# Codificación(Encoding) Transforma variables no numéricas (por ejemplo, sí/no, categorías) y  las  transforma en valores numéricos
#                         para que puedan ser utilizadas por modelos de aprendizaje automático. 
#                         Este paso permite inspeccionar variables categóricas antes de su codificación.Nota: Se dejan algunos terminos en Ingles para facilitar la investigacion

 

#Paso 1 —   Codificacion a valores Binarios
            #Para vvalores Binarios "yes" = 1, "no" =  0
            #df["diabetes"] → select the column #    .map({...}) → replace values using a rule   #    "yes" becomes 1, "no" becomes 0
            #After this: the column is numeric  values are 0 or 1 ready for ML models later
            #You should now see: only 0 and 1 no "yes" or "no"
print("ANTES del encoding:")
#df[["diabetes", "smoker"]].head()
display(df[["diabetes", "smoker"]].head())
#Las columnas categóricas se visualizan antes y después del proceso de codificación para verificar 
#la correcta transformación de valores textuales en representaciones numéricas.

# Codificación(Encouding) 
df["diabetes"] = df["diabetes"].map({"yes": 1, "no": 0}) # Ocurre la codificación (encoding)..map({...}) reemplaza valores "yes" → 1 "no" → 0
df["smoker"] = df["smoker"].map({"yes": 1, "no": 0})

print("DESPUÉS del encoding:")
#df[["diabetes", "smoker"]].head()
display(df[["diabetes", "smoker"]].head())




```

## Pipeline meaning:
Esto implementa la fase de normalization descrita en el artículo.
Tras el escalado, los valores representan la distancia de cada observación respecto a la media en unidades de desviación estándar, por lo que pueden tomar valores negativos o superiores a uno sin que ello suponga un error.
Aclaración  

Si el valor escalado es −1.0 → está 1 desviación estándar por debajo de la media.
Si el valor escalado es −1.5 → está 1.5 desviaciones estándar por debajo de la media.

Los valores escalados se interpretan en relación con la media y la desviación estándar del conjunto de datos analizado, por lo que representan posiciones relativas dentro de la muestra.

StandardScaler changes the ruler, not the ordering.
Negative values are below the dataset mean, not “wrong”.
Scaling statistics are dataset‑relative, not universal truths.
Rounding is for readability, not learning.
Pipelines are about flow, not libraries.




El escalado se realiza por columna, de forma independiente para cada variable, garantizando que las comparaciones se efectúen entre características homogéneas y no entre magnitudes distintas dentro de una misma observación.

#Si quitas fit_transform() → no hay escalado
#Si quitas columns=... → hay escalado, pero no sabes qué columna es cuál
 

#✅ El escalado ocurre en fit_transform()
#✅ Es por columna, no por fila
#✅ columns=... solo pone nombres

 

```python
#
# Scaling / Escalado
#    (ES): El escalado es una técnica de preprocesamiento que transforma variables numéricas para que estén en la misma escala,
#       evitando que unas dominen a otras solo por su magnitud.
#    (EN): Scaling ensures that numerical features contribute equally to machine learning models by putting them on comparable scales.
#

# Paso 1 — Importa las librerias necesarias para la normalizacion
from sklearn.preprocessing import StandardScaler

# Paso 2 — Identificar columnas numéricas — Identify which columns should be scaled
#      Solo las variables numéricas se pueden escalar
#      Variables categóricas o texto no se  pasan por el scaler

numeric_df = df.select_dtypes(include="number") #todas las columnas numéricas del DataFrame.



# Paso 3 — Excluye variables de estado (binary states) /  Exclude binary state columns 
continuous_df = numeric_df.drop(columns=["diabetes", "smoker"]) #Quita variables binarias de estado {(0/1),(presencia/ausencia)} del escalado 
                                                                # Binary variables represent states, not magnitudes. Dropping the state columns for the scaler only.
                                                 
                                                                
# Paso 4 — Aplicar El Escalado  StandardScaler
#      Es una técnica de normalización que centra los datos en 0, ajusta la desviación estándar a 1
#      escala cada columna de forma independiente, por columnas, toma una columna a la vez, calcula la media y desviación estándar 
#      de ESA columna,transforma solo esa columna luego la siguiente.Ejemplo: colesterol → se estandariza usando media y desviación del colesterol
#      y devuelve un array NumPy, los centra los datos en 0, ajusta la desviación estándar a 1 
#      Ejemplo del Escalado/normalizacion:  Cuántas desviaciones estándar está ese valor respecto a la media:
#                        0 → valor promedio,  
#                        negativo → por debajo del promedio
#                        positivo → por encima del promedio
#                 Un valor escalado de:
#                       +1 → una desviación estándar por encima de la media
#                       +2 → dos desviaciones estándar
#                       +3 → muy lejos de la media (posible outlier)
#                     

scaler = StandardScaler()
scaled_array = scaler.fit_transform(continuous_df) # escalado se hace por columna, no por fila 
#print(scaled_array)


# Paso 5 — Volver a forma tabular /  Put scaled data back into a DataFrame
           # Lo convertimos en una tabla para que se leible / We convert it back into a table so it’s readable.
    
scaled_df = pd.DataFrame(
    scaled_array,columns=continuous_df.columns,  index=df.index)

 
  
# Paso 5 — Inspección del resultado  Inspect the result  
           #Tras el proceso de escalado, los valores no se redondean para preservar la precisión necesaria 
           #en el entrenamiento del modelo, utilizándose el redondeo únicamente con fines de visualización.
#scaled_df.head() 
scaled_df.round(2).head() # Se muestran redondeados para facilitar la lectura pero internamente mantienen su precisión

```

```python
# 
# Paso ML 1 Crear el TARGET (y)
#         Antes de entrenar cualquier modelo, se define explícitamente qué significa estar en riesgo.
#         Se crea la variable objetivo "at_risk" que es lo que el modelo va a aprender usando valores ya escalados 
#         Se  define qué significa "Estar en riesgo" que en este caso se se escogió el valor 0,8
#         Valor escalado:
#               0 → valor medio del dataset
#               0.8 → 0.8 desviaciones estándar por encima de la media
#         Se  crea primero "la verdad (y)", luego se prepara lo que el modelo va a mirar (X).
#         Este target es: es sintético, es didáctico y se basa en valores escalados

# Un paciente se considera “en riesgo” solo si ambas variables están elevadas simultáneamente.Ejemplo colesterol + Lipoproteina a 
# Z‑score(colesterol)  ->  Este valor está X desviaciones estándar por encima o por debajo de la media del dataset. 
#      Interpretacion de los Valores
#            Número (magnitud) → lectura (intensidad) -> magnitud (0.3, 0.8, 1.5) → intensidad(Ligeramente,Claramente, Muy por encima del promedio) 
#            Magnitud (el número) Colesterol = 0.3
#            Valor = 0.0  Este paciente tiene un colesterol exactamente en la media del conjunto de datos.
#            Valor = 0.8  Este paciente tiene un colesterol 0.8 desviaciones estándar por encima de la media, "Claramente por encima del promedio"
df["at_risk"] = (
    (scaled_df["total_cholesterol_mg_dl"] > 0.8) &      # AND(&) = condicion1 & condicion2 = True & True → True
    (scaled_df["lipoprotein_a_mg_dl"] > 0.8)           
).astype(int) # Convierte: True → 1, False → 0 
# lo que el modelo vera :  
#            at_risk = 1  Paciente en Riesgo 
#            at_risk = 0  Sin riesgo
```
