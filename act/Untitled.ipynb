```python
Root Dictionary 
   └── "features" (The List of all stations)
         ├── [Station 0]
         │     ├── "attributes" (The text data: name, no2, pm10)
         │     └── "geometry"   (The map coordinates: x, y)
         └── [Station 1]
               ├── "attributes" (...)
               └── "geometry"   (...)
```

```python
import requests

url = "https://geoportal.valencia.es/server/rest/services/OPENDATA/MedioAmbiente/MapServer/156/query"
params = {"where": "1=1", "f": "json", "outFields": "*"}

# Fetch the root dictionary
data = requests.get(url, params=params).json()

# 1. Step into the "features" list you identified
features_list = data["features"]

# 2. Grab the very first station object in that list (index 0)
first_station = features_list[0]

# 3. Separate the attributes from the geometry
station_attributes = first_station["attributes"]
station_geometry = first_station["geometry"]

# Now print them cleanly to see your work match the screen
print(f"Station Name: {station_attributes['nombre']}")
print(f"Coordinates: X={station_geometry['x']}, Y={station_geometry['y']}")
```

```python
import requests

# 1. Target URL from your browser screenshot with parameters
url = "https://geoportal.valencia.es/server/rest/services/OPENDATA/MedioAmbiente/MapServer/156/query"
params = {
    "where": "1=1",
    "outFields": "nombre,direccion,so2,no2,pm10,calidad_am",
    "f": "json"
}

# 2. Execute request and format as JSON
response = requests.get(url, params=params)
data = response.json()

# 3. Access the 'features' list to extract station records
stations = data.get("features", [])

print(f"Found {len(stations)} air pollution monitoring stations:\n")
print(f"{'Station Name':<15} | {'NO2 Level':<10} | {'PM10 Level':<10} | {'Air Quality'}")
print("-" * 60)

# 4. Loop through each station and print its attributes
for station in stations:
    attribs = station.get("attributes", {})
    
    name = attribs.get("nombre", "Unknown")
    no2 = attribs.get("no2", "N/A")
    pm10 = attribs.get("pm10", "N/A")
    quality = attribs.get("calidad_am", "Unknown")
    
    print(f"{name:<15} | {str(no2):<10} | {str(pm10):<10} | {quality}")
```
