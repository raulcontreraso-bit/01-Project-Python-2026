```python
from time import sleep
import os
import requests

# 1. Base Setup & Configuration
base_url = "https://geoportal.valencia.es/server/rest/services/OPENDATA/MedioAmbiente/MapServer/156/query"

# Ensure the storage folder exists on your drive
if not os.path.exists('res'):
    os.makedirs('res')

# Time intervals (Set to 1 second each for quick notebook testing!)
current_time = 0
sleep_time = 1
total_time = 1

max_records_per_request = 100

# 2. Reusable Request Function
def make_request(url, params):
    response = requests.get(url, params=params)
    if response:
        return response.json()
    else:
        raise Exception("Error Downloading JSON")

# 3. Get the Baseline Benchmark (Counting rows before starting)
query_params = {
    "where": "1=1",
    "outFields": "*",
    "f": "json"
}
raw_data = make_request(base_url, params=query_params)
n_stations = len(raw_data['features'])

# 4. The Data Collection Loop
while current_time < total_time:
    offset = 0
    pollution_list = []

    # Inner loop to handle records matching our benchmark
    while offset < n_stations:
        params = {
            "where": "1=1",
            "outFields": "*",
            "f": "json"
        }

        # Step into the correct dictionary key ('features')
        sub_list = make_request(base_url, params)['features']
        
        # Use our "fancy loop" to extract only the inner data measurements
        clean_rows = [station['attributes'] for station in sub_list]
        pollution_list += clean_rows

        # Advance the offset to safely break the inner loop
        offset += max_records_per_request

    # Print the clean Python result to the screen
    print(f"Current Time = {current_time}, Records Processed = {len(pollution_list)}")
    print("Sample Data gathered:", pollution_list[0])

    # Advance the clock and pause
    current_time += sleep_time
    sleep(sleep_time)
```
