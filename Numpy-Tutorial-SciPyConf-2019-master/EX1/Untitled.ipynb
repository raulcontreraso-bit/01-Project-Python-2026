🛠️ Phase 1: Core Python FoundationsYour goal here is to understand syntax, scoping, and pythonic logic patterns before introducing complex external math libraries.📚 Learning ResourcesThe Core Video: Matt Davis - Introduction to Python (SciPy) (approx. 2.3 hours)The Companion Guide: Use the python_core_crash_course.md currently open in your sidebar to track and verify your data type definitions.🎯 Milestone Checklist[ ] You can declare lists and dictionaries without syntax errors.[ ] You can build a function using parameters, run its logic, and return the computed results.[ ] You understand that changing a variable b = a points both nametags to the same object in memory.🥊 Shadow Boxing Sparring ExerciseWrite a function that accepts a list of numbers and returns only the even numbers. Intentionally run it with a dictionary instead of a list to observe and analyze the resulting TypeError traceback.🛠️ Phase 2: NumPy (The Grid Engine)NumPy replaces slow standard Python loops with vectorized, contiguous blocks of numbers managed directly in RAM.📚 Learning ResourcesThe Core Video: Alex Chabot-Leclerc - NumPy Tutorial (SciPy) (approx. 2 hours)Official Guide: NumPy: The Absolute Basics for Beginners (15-minute quickstart)🎯 Milestone Checklist[ ] You understand the physical difference between a 1D vector and a 2D matrix in memory.[ ] You can slice a grid of data using grid[row_slice, column_slice].[ ] You know that Axis 0 means working "down" (vertically) and Axis 1 means working "across" (horizontally).🥊 Shadow Boxing Sparring ExerciseCreate a 3x3 array containing numbers 1 through 9. Try to write a slice command that extracts the 2x2 square in the bottom right corner:# Expected Output:
[[5, 6],
 [8, 9]]
If you get index boundary errors, step back and analyze your coordinate boundaries!🛠️ Phase 3: Pandas (The Data Frame Machine)Pandas is built directly on top of NumPy. Instead of raw numeric grids, it handles structured, labeled tables (DataFrames) exactly like a supercharged Excel spreadsheet inside your code.📚 Learning ResourcesThe Fast-Track Read: Official 10 Minutes to pandas TutorialThe Full Tutorial Video: Daniel Chen - Pandas for Data Analysis (SciPy) (approx. 3.5 hours)🎯 Milestone Checklist[ ] You can load a raw .csv file into a DataFrame using pd.read_csv().[ ] You can filter rows instantly using boolean expressions (e.g., df[df['age'] > 30]).[ ] You can group data by a category and calculate averages (e.g., .groupby('City')['Salary'].mean()).🥊 Shadow Boxing Sparring ExerciseOpen the official "10 Minutes to pandas" guide, open an empty Jupyter Notebook next to it, and manually copy-type and execute every single cell. If a cell fails, check your imports or the dataset's index.🧠 Master Rules of the Learning MindsetThe "Copy-Paste" Ban: Never use copy-paste while following these tutorials. Typing the letters physically is what programs your fingers and neural pathways to remember the syntax.Celebrate the Errors: When Python throws a massive red box of errors at you, do not get frustrated. It's the compiler talking to you. Read the very last line of the error—it always tells you the exact symptom.Build Immediately: As soon as you finish the Pandas "10 Minutes" tutorial, don't watch another video. Find a free dataset on something you love (sports stats, movies, space data), load it, and try to sort, filter, and summarize it on your own.
