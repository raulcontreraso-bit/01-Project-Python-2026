```python

import time
import requests      #tool to talk to websites, like opening a browser automatically
import pandas as pd  # tool to create tables (Excel‑like), like having Excel inside Python
                    #Without requests → ❌ cannot get API data
                    #Without pandas → ❌ cannot build the dataset
from pyproj import Transformer

transformer = Transformer.from_crs("EPSG:32630", "EPSG:4326", always_xy=True)


# “Give me pollution data from this specific dataset”
server_url = "https://geoportal.valencia.es/server/rest/services/OPENDATA/"   #  tell Python where to go to get the data.  address building ✅
base_url = server_url + "MedioAmbiente/MapServer/156/query"                   # address specific room ✅



#Tell the API what format and how much data we want.
params = {
    'f': 'json',                # Send the data in JSON format, JSON = structured data (lists + dictionaries)
    'outFields': '*',           # Give me ALL columns, means EVERYTHING
    'where': '1=1'              #  “Give me ALL rows”, 1=1 is always true, so nothing is filtered
}

# dictionary 
# keys → 'f', 'outFields', 'where'
# values → 'json', '*', '1=1'
# is exactly like: Pythonitem = {"name": "A", "pollution": 40}

response = requests.get(base_url, params=params)  #  API call
#requests.get(...) 👉 sends a request to the server
# base_url 👉 where to go
# params=params 👉 what we want
#response 👉 the server’s answer

#🔍 What this line does (simple + exact)
#
#requests.get(...) → sends the request
#base_url → where to go ✅
#params=params → what data you want ✅

#import json
#print(json.dumps(data, indent=2))

data = response.json()  #raw response → structured Python (list + dictionaries)
#print(data.keys())
#print(type(data["features"]))
#print(data["features"][0])
#print(data["features"][0].keys())

features = data["features"]   # ✅ THIS WAS MISSING
print(len(features))
#print(features)


#df = pd.DataFrame()

#for feature in features:
#    transformer = Transformer.from_crs("EPSG:32630", "EPSG:4326", always_xy=True)

    
#    row = pd.Series(feature["attributes"])   #      
    # then add coordinates
#    row["x"] = feature["geometry"]["x"]
#    row["y"] = feature["geometry"]["y"]
    #🧠 Think of it like Excel
    #row = one row in Excel
    #row["x"] = ... = adding a new column cell
    #print(row)
    #print(row.keys())
#   df = pd.concat([df, row.to_frame().T], ignore_index=True)    
#   print(feature.keys())
#   print(feature["attributes"].keys())

# print(len(df))

# print(features[0]["attributes"]["fecha_carg"])
# print(df.iloc[0]["fecha_carg"])
#👉 Compare:
#
#API value
#stored value
print()

#print(data.keys())
#print(type(data["features"]))
#print(data["features"][0])
print()
#print(data["features"][0]["attributes"])
#print(data["features"][0]["attributes"].keys())



#Why "features" works in THIS API
#Because when you inspect:
#Pythondata["features"][0]Show more lines
#👉 you see a repeated pattern like:
#{"attributes": {...}, "geometry": {...}}

#That repetition tells you:
#👉 “this is a list of rows”

#feature["attributes"]
#station   pollution   fecha_carg
#A         40          2024-05-01



df = pd.DataFrame()

#for i in range(24): 
for i in range(2):
    
    response = requests.get(base_url, params=params)
    data = response.json()
    features = data["features"]

    for feature in features:
        row = pd.Series(feature["attributes"])

        x = feature["geometry"]["x"]
        y = feature["geometry"]["y"]

        lon, lat = transformer.transform(x, y)

        row["lon"] = lon
        row["lat"] = lat

        df = pd.concat([df, row.to_frame().T], ignore_index=True)

    #time.sleep(3600)   # wait 1 hour
    time.sleep(2)      # wait 2 seconds







df["fecha_carg"] = pd.to_datetime(df["fecha_carg"], unit='ms')

print(df.head())

print (df)



# after building df
df.to_csv("valencia_pollution_rco.csv", index=False) #pandas:Save this table as a CSV file, index=False: Do NOT save those row numbers in the file
```
